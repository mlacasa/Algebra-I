# La Fórmula de Grassmann: Estructura y Dimensión de Subespacios Vectoriales

## Introducción

La fórmula de Grassmann, nombrada en honor al matemático alemán Hermann Grassmann (1809-1877), establece una relación fundamental entre las dimensiones de la suma y la intersección de dos subespacios vectoriales. Esta relación es esencial en álgebra lineal y tiene importantes aplicaciones en geometría algebraica, teoría de códigos, y análisis funcional.

La fórmula se expresa como:

$$\dim(U + V) = \dim(U) + \dim(V) - \dim(U \cap V)$$

donde $U$ y $V$ son subespacios vectoriales de un espacio vectorial.

## Definiciones Preliminares

Antes de abordar la demostración, recordemos algunas definiciones esenciales:

### Suma de Subespacios
Dados dos subespacios $U$ y $V$ de un espacio vectorial $W$, la suma $U + V$ se define como:

$$U + V = \{u + v \mid u \in U, v \in V\}$$

Esto es, el conjunto de todos los vectores que pueden expresarse como suma de un vector en $U$ y un vector en $V$.

### Intersección de Subespacios
La intersección $U \cap V$ es el conjunto de todos los vectores que pertenecen simultáneamente a $U$ y $V$:

$$U \cap V = \{w \mid w \in U \text{ y } w \in V\}$$

### Dimensión de un Subespacio
La dimensión de un subespacio vectorial es el número máximo de vectores linealmente independientes que contiene, o equivalentemente, el tamaño de cualquier base del subespacio.

## Demostración de la Fórmula de Grassmann

Presentamos una demostración rigurosa de la fórmula de Grassmann utilizando bases de los subespacios involucrados.

### Paso 1: Construcción de una Base para $U \cap V$

Sea $\{w_1, w_2, \ldots, w_k\}$ una base de $U \cap V$, donde $k = \dim(U \cap V)$.

### Paso 2: Extensión a Bases de $U$ y $V$

Dado que $U \cap V \subseteq U$, podemos extender la base de $U \cap V$ a una base de $U$ añadiendo vectores adecuados:

$$\{w_1, w_2, \ldots, w_k, u_1, u_2, \ldots, u_m\}$$

donde $\dim(U) = k + m$.

De manera similar, extendemos la base de $U \cap V$ a una base de $V$:

$$\{w_1, w_2, \ldots, w_k, v_1, v_2, \ldots, v_n\}$$

donde $\dim(V) = k + n$.

### Paso 3: Construcción de una Base para $U + V$

Afirmamos que el conjunto:

$$\{w_1, w_2, \ldots, w_k, u_1, u_2, \ldots, u_m, v_1, v_2, \ldots, v_n\}$$

forma una base para $U + V$.

Para demostrar esto, debemos probar que:

1. Este conjunto genera $U + V$
2. Los vectores en este conjunto son linealmente independientes

#### 3.1 Demostración de que el conjunto genera $U + V$

Cualquier vector $z \in U + V$ puede escribirse como $z = u + v$ donde $u \in U$ y $v \in V$.

Como $u \in U$, podemos expresar $u$ en términos de la base de $U$:

$$u = \sum_{i=1}^{k} \alpha_i w_i + \sum_{j=1}^{m} \beta_j u_j$$

De manera similar, para $v \in V$:

$$v = \sum_{i=1}^{k} \gamma_i w_i + \sum_{l=1}^{n} \delta_l v_l$$

Combinando estas expresiones:

$$z = u + v = \sum_{i=1}^{k} (\alpha_i + \gamma_i) w_i + \sum_{j=1}^{m} \beta_j u_j + \sum_{l=1}^{n} \delta_l v_l$$

Esto demuestra que nuestro conjunto genera $U + V$.

#### 3.2 Demostración de la independencia lineal

Consideremos una combinación lineal que se anula:

$$\sum_{i=1}^{k} \alpha_i w_i + \sum_{j=1}^{m} \beta_j u_j + \sum_{l=1}^{n} \gamma_l v_l = 0$$

Reordenando:

$$\sum_{j=1}^{m} \beta_j u_j = -\sum_{i=1}^{k} \alpha_i w_i - \sum_{l=1}^{n} \gamma_l v_l$$

El lado izquierdo pertenece a $U$. En el lado derecho, los términos con $w_i$ también pertenecen a $U$. Por lo tanto:

$$\sum_{l=1}^{n} \gamma_l v_l \in U$$

Pero como los vectores $v_l$ se eligieron específicamente para complementar la base de $U \cap V$ en $V$, la única manera de que esta combinación lineal esté en $U$ es si todos los $\gamma_l = 0$.

Por un argumento similar, todos los $\beta_j = 0$.

Finalmente, como los $w_i$ son linealmente independientes (forman una base de $U \cap V$), debemos tener $\alpha_i = 0$ para todo $i$.

Esto prueba que los vectores son linealmente independientes.

### Paso 4: Contar la Dimensión

Como hemos construido una base para $U + V$ con exactamente $k + m + n$ vectores, concluimos:

$$\dim(U + V) = k + m + n$$

Pero sabemos que:
- $\dim(U) = k + m$
- $\dim(V) = k + n$
- $\dim(U \cap V) = k$

Sustituyendo:

$$\dim(U + V) = \dim(U) + \dim(V) - \dim(U \cap V)$$

Lo que completa la demostración.

## Interpretación Geométrica

La fórmula de Grassmann tiene una interpretación geométrica intuitiva:

- Si pensamos en $\dim(U)$ y $\dim(V)$ como "volúmenes" de los respectivos subespacios, al sumarlos directamente, estamos contando dos veces la parte común (la intersección).
- Para obtener el "volumen" correcto de la suma, necesitamos restar una vez la intersección.

Esta interpretación es análoga al principio de inclusión-exclusión en teoría de conjuntos:

$$|A \cup B| = |A| + |B| - |A \cap B|$$

## Casos Particulares y Ejemplos

### Caso 1: Subespacios con Intersección Trivial

Si $U \cap V = \{0\}$ (solo contienen el vector cero en común), entonces $\dim(U \cap V) = 0$ y la fórmula se reduce a:

$$\dim(U + V) = \dim(U) + \dim(V)$$

En este caso, decimos que $U$ y $V$ son subespacios complementarios.

**Ejemplo:** En $\mathbb{R}^3$, sean $U = \text{span}\{(1,0,0)\}$ (eje $x$) y $V = \text{span}\{(0,1,0), (0,0,1)\}$ (plano $yz$). Entonces:
- $\dim(U) = 1$
- $\dim(V) = 2$
- $\dim(U \cap V) = 0$
- $\dim(U + V) = 1 + 2 - 0 = 3$

Esto corresponde al hecho de que el eje $x$ y el plano $yz$ generan todo $\mathbb{R}^3$.

### Caso 2: Un Subespacio Contenido en el Otro

Si $U \subseteq V$, entonces $U \cap V = U$ y la fórmula da:

$$\dim(U + V) = \dim(U) + \dim(V) - \dim(U) = \dim(V)$$

Lo cual es correcto ya que $U + V = V$ en este caso.

**Ejemplo:** Sean $U = \text{span}\{(1,0,0)\}$ (eje $x$) y $V = \text{span}\{(1,0,0), (0,1,0)\}$ (plano $xy$). Entonces:
- $\dim(U) = 1$
- $\dim(V) = 2$
- $\dim(U \cap V) = 1$
- $\dim(U + V) = 1 + 2 - 1 = 2$

Esto confirma que $U + V = V$, ya que $U \subseteq V$.

## Aplicaciones de la Fórmula de Grassmann

La fórmula de Grassmann tiene numerosas aplicaciones en:

1. **Teoría de Códigos Lineales**: Para calcular la dimensión de códigos compuestos.
2. **Geometría Diferencial**: Para estudiar intersecciones de variedades.
3. **Teoría de Representación**: Para analizar representaciones de grupos y álgebras.
4. **Ecuaciones Diferenciales**: Para analizar espacios de soluciones.
5. **Teoría de Matrices**: Para calcular rangos de sumas de matrices.

## Generalización a Múltiples Subespacios

La fórmula de Grassmann puede generalizarse a más de dos subespacios, aunque las fórmulas resultan más complejas e involucran términos de alternancia de sumas y restas similar al principio de inclusión-exclusión.

## Conclusión

La fórmula de Grassmann representa uno de los resultados fundamentales del álgebra lineal que relaciona la estructura de subespacios vectoriales. Su elegancia radica en la forma en que captura precisamente cómo las dimensiones de subespacios se relacionan a través de sus operaciones básicas de suma e intersección, proporcionando una herramienta poderosa para el análisis y la comprensión de espacios vectoriales.

```python
import numpy as np
from numpy.linalg import matrix_rank
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def dimensiones_subespacios(U_base, V_base, verbose=True):
    """
    Calcula las dimensiones de U, V, U+V y U∩V y verifica la fórmula de Grassmann:
    dim(U + V) = dim(U) + dim(V) - dim(U ∩ V)

    Args:
        U_base: Lista de vectores que forman la base del subespacio U
        V_base: Lista de vectores que forman la base del subespacio V
        verbose: Si True, imprime los resultados paso a paso

    Returns:
        Un diccionario con las dimensiones calculadas y la verificación de la fórmula
    """
    # Convertir las bases a matrices (cada fila es un vector de la base)
    U_matriz = np.array(U_base)
    V_matriz = np.array(V_base)

    # Dimensión de U y V (es el rango de las matrices formadas por sus bases)
    dim_U = matrix_rank(U_matriz)
    dim_V = matrix_rank(V_matriz)

    if verbose:
        print(f"Base de U: \n{U_matriz}")
        print(f"Base de V: \n{V_matriz}")
        print(f"Dimensión de U: {dim_U}")
        print(f"Dimensión de V: {dim_V}")

    # Para U + V, concatenamos las bases y calculamos el rango
    UV_suma = np.vstack((U_matriz, V_matriz))
    dim_U_mas_V = matrix_rank(UV_suma)

    if verbose:
        print(f"\nVectores de U y V juntos: \n{UV_suma}")
        print(f"Dimensión de U + V: {dim_U_mas_V}")

    # Calculamos la dimensión de U ∩ V usando la fórmula de Grassmann
    dim_U_intersect_V = dim_U + dim_V - dim_U_mas_V

    if verbose:
        print(f"\nDimensión de U ∩ V (calculada con Grassmann): {dim_U_intersect_V}")

    # Verificamos la fórmula de Grassmann
    lado_izquierdo = dim_U_mas_V
    lado_derecho = dim_U + dim_V - dim_U_intersect_V

    if verbose:
        print("\nVerificación de la fórmula de Grassmann:")
        print(f"dim(U + V) = {lado_izquierdo}")
        print(f"dim(U) + dim(V) - dim(U ∩ V) = {dim_U} + {dim_V} - {dim_U_intersect_V} = {lado_derecho}")
        print(f"¿Se cumple la fórmula? {lado_izquierdo == lado_derecho}")

    # Intentamos encontrar una base para U ∩ V
    # Esto es más complejo y requiere resolver sistemas de ecuaciones lineales
    matriz_U_expandida = np.zeros((U_matriz.shape[0], max(U_matriz.shape[1], V_matriz.shape[1])))
    matriz_V_expandida = np.zeros((V_matriz.shape[0], max(U_matriz.shape[1], V_matriz.shape[1])))

    matriz_U_expandida[:, :U_matriz.shape[1]] = U_matriz
    matriz_V_expandida[:, :V_matriz.shape[1]] = V_matriz

    # Construimos el sistema de ecuaciones aumentado
    A = np.vstack((matriz_U_expandida, matriz_V_expandida))
    rango_A = matrix_rank(A)

    n = A.shape[1]  # número de variables
    dim_nucleo = n - rango_A  # dimensión del núcleo de A

    if verbose:
        print("\nMétodo alternativo para calcular la intersección:")
        print(f"Dimensión del núcleo de la matriz combinada: {dim_nucleo}")
        if dim_nucleo == dim_U_intersect_V:
            print("Esto coincide con la dimensión calculada usando la fórmula de Grassmann.")
        else:
            print("Advertencia: Los resultados no coinciden, podría haber un error numérico.")

    return {
        "dim_U": dim_U,
        "dim_V": dim_V,
        "dim_U_mas_V": dim_U_mas_V,
        "dim_U_intersect_V": dim_U_intersect_V,
        "verificacion": lado_izquierdo == lado_derecho
    }

def visualizar_subespacios_R3(U_base, V_base, num_puntos=10):
    """
    Visualiza subespacios vectoriales en R³ para los casos donde las bases están en R³.

    Args:
        U_base: Lista de vectores que forman la base del subespacio U
        V_base: Lista de vectores que forman la base del subespacio V
        num_puntos: Número de puntos a generar para la visualización
    """
    # Verificar que estamos en R³
    if len(U_base[0]) != 3 or len(V_base[0]) != 3:
        print("Esta función solo visualiza subespacios en R³")
        return

    # Convertir a matrices numpy
    U_matriz = np.array(U_base)
    V_matriz = np.array(V_base)

    # Crear figura 3D
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')

    # Generar puntos del subespacio U
    dim_U = matrix_rank(U_matriz)
    if dim_U == 1:  # Línea
        t = np.linspace(-2, 2, num_puntos)
        U_puntos = np.outer(t, U_base[0])
    elif dim_U == 2:  # Plano
        t1, t2 = np.meshgrid(np.linspace(-1, 1, num_puntos), np.linspace(-1, 1, num_puntos))
        U_puntos = np.outer(t1.flatten(), U_base[0]) + np.outer(t2.flatten(), U_base[1])
        U_puntos = U_puntos.reshape(num_puntos, num_puntos, 3)
    elif dim_U == 3:  # Espacio completo
        print("U es todo R³, no se puede visualizar efectivamente")

    # Generar puntos del subespacio V
    dim_V = matrix_rank(V_matriz)
    if dim_V == 1:  # Línea
        s = np.linspace(-2, 2, num_puntos)
        V_puntos = np.outer(s, V_base[0])
    elif dim_V == 2:  # Plano
        s1, s2 = np.meshgrid(np.linspace(-1, 1, num_puntos), np.linspace(-1, 1, num_puntos))
        V_puntos = np.outer(s1.flatten(), V_base[0]) + np.outer(s2.flatten(), V_base[1])
        V_puntos = V_puntos.reshape(num_puntos, num_puntos, 3)
    elif dim_V == 3:  # Espacio completo
        print("V es todo R³, no se puede visualizar efectivamente")

    # Dibujar los subespacios
    if dim_U == 1:
        ax.scatter(U_puntos[:, 0], U_puntos[:, 1], U_puntos[:, 2], color='red', alpha=0.5, label='U (línea)')
    elif dim_U == 2:
        ax.plot_surface(U_puntos[:, :, 0], U_puntos[:, :, 1], U_puntos[:, :, 2], color='red', alpha=0.3, label='U (plano)')

    if dim_V == 1:
        ax.scatter(V_puntos[:, 0], V_puntos[:, 1], V_puntos[:, 2], color='blue', alpha=0.5, label='V (línea)')
    elif dim_V == 2:
        ax.plot_surface(V_puntos[:, :, 0], V_puntos[:, :, 1], V_puntos[:, :, 2], color='blue', alpha=0.3, label='V (plano)')

    # Configuraciones del gráfico
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Visualización de los subespacios U y V en R³')

    # Establecer límites
    limite = 3
    ax.set_xlim([-limite, limite])
    ax.set_ylim([-limite, limite])
    ax.set_zlim([-limite, limite])

    # Agregar origen
    ax.scatter([0], [0], [0], color='black', s=50)

    plt.show()

# Ejemplo 1: Dos líneas que se intersecan en el origen en R³
print("EJEMPLO 1: Dos líneas en R³ que se intersecan en el origen")
U_base_1 = [[1, 0, 0]]  # Línea a lo largo del eje X
V_base_1 = [[0, 1, 0]]  # Línea a lo largo del eje Y
dimensiones_subespacios(U_base_1, V_base_1)
visualizar_subespacios_R3(U_base_1, V_base_1)

# Ejemplo 2: Una línea y un plano que contiene esa línea en R³
print("\nEJEMPLO 2: Una línea y un plano que contiene esa línea en R³")
U_base_2 = [[1, 0, 0]]  # Línea a lo largo del eje X
V_base_2 = [[1, 0, 0], [0, 1, 0]]  # Plano XY
dimensiones_subespacios(U_base_2, V_base_2)
visualizar_subespacios_R3(U_base_2, V_base_2)

# Ejemplo 3: Dos planos que se intersecan en una línea en R³
print("\nEJEMPLO 3: Dos planos que se intersecan en una línea en R³")
U_base_3 = [[1, 0, 0], [0, 1, 0]]  # Plano XY
V_base_3 = [[1, 0, 0], [0, 0, 1]]  # Plano XZ
dimensiones_subespacios(U_base_3, V_base_3)
visualizar_subespacios_R3(U_base_3, V_base_3)

# Ejemplo 4: Dos subespacios en R⁴
print("\nEJEMPLO 4: Dos subespacios en R⁴")
U_base_4 = [[1, 0, 0, 0], [0, 1, 0, 0]]  # Subespacio generado por e₁ y e₂
V_base_4 = [[0, 0, 1, 0], [0, 0, 0, 1]]  # Subespacio generado por e₃ y e₄
dimensiones_subespacios(U_base_4, V_base_4)
# No podemos visualizar este ejemplo en R⁴

# Ejemplo 5: Dos subespacios con intersección no trivial en R⁴
print("\nEJEMPLO 5: Dos subespacios con intersección no trivial en R⁴")
U_base_5 = [[1, 0, 1, 0], [0, 1, 0, 1]]
V_base_5 = [[1, 0, 1, 0], [1, 1, 1, 1]]
dimensiones_subespacios(U_base_5, V_base_5)
# No podemos visualizar este ejemplo en R⁴
```
