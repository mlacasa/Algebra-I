# Cálculo de la Base Dual en R³

## Planteamiento del problema

Tenemos la base de R³:
$$\mathcal{B} = \{u_1 = (1,1,1), u_2 = (1,1,0), u_3 = (1,0,0)\}$$

Queremos calcular la base dual $\mathcal{B}^* = \{u_1^*, u_2^*, u_3^*\}$, que debe cumplir:
$$u_i^*(u_j) = \delta_{ij} \quad \text{para } 1 \leq i,j \leq 3$$

Donde $\delta_{ij}$ es la delta de Kronecker:
$$\delta_{ij} = \begin{cases}
1 & \text{si } i = j \\
0 & \text{si } i \neq j
\end{cases}$$

## Teoría sobre la base dual

La base dual de una base de un espacio vectorial es una base del espacio dual que satisface la condición $u_i^*(u_j) = \delta_{ij}$. En R³, podemos pensar en los elementos $u_i^*$ como formas lineales (funcionales lineales) que mapean vectores a escalares.

Dado que estamos en R³, podemos representar los elementos de la base dual como vectores en R³ tal que:
$$u_i^*(u_j) = \langle u_i^*, u_j \rangle = \delta_{ij}$$

Donde $\langle \cdot, \cdot \rangle$ denota el producto escalar estándar.

## Resolución

Para encontrar la base dual, necesitamos resolver un sistema de ecuaciones lineales. Para cada $u_i^*$, debemos satisfacer las condiciones:
- $\langle u_i^*, u_1 \rangle = \delta_{i1}$
- $\langle u_i^*, u_2 \rangle = \delta_{i2}$
- $\langle u_i^*, u_3 \rangle = \delta_{i3}$

### Encontrando $u_1^*$

Si $u_1^* = (a, b, c)$, debemos cumplir:
- $\langle u_1^*, u_1 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 1 = 1$
- $\langle u_1^*, u_2 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 0 = 0$
- $\langle u_1^*, u_3 \rangle = a \cdot 1 + b \cdot 0 + c \cdot 0 = 0$

Esto nos da el sistema:
- $a + b + c = 1$
- $a + b = 0$
- $a = 0$

Resolviendo:
- De la tercera ecuación: $a = 0$
- Sustituyendo en la segunda: $0 + b = 0$, por lo que $b = 0$
- Sustituyendo en la primera: $0 + 0 + c = 1$, por lo que $c = 1$

Por lo tanto, $u_1^* = (0, 0, 1)$.

### Encontrando $u_2^*$

Si $u_2^* = (a, b, c)$, debemos cumplir:
- $\langle u_2^*, u_1 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 1 = 0$
- $\langle u_2^*, u_2 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 0 = 1$
- $\langle u_2^*, u_3 \rangle = a \cdot 1 + b \cdot 0 + c \cdot 0 = 0$

Esto nos da el sistema:
- $a + b + c = 0$
- $a + b = 1$
- $a = 0$

Resolviendo:
- De la tercera ecuación: $a = 0$
- Sustituyendo en la segunda: $0 + b = 1$, por lo que $b = 1$
- Sustituyendo en la primera: $0 + 1 + c = 0$, por lo que $c = -1$

Por lo tanto, $u_2^* = (0, 1, -1)$.

### Encontrando $u_3^*$

Si $u_3^* = (a, b, c)$, debemos cumplir:
- $\langle u_3^*, u_1 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 1 = 0$
- $\langle u_3^*, u_2 \rangle = a \cdot 1 + b \cdot 1 + c \cdot 0 = 0$
- $\langle u_3^*, u_3 \rangle = a \cdot 1 + b \cdot 0 + c \cdot 0 = 1$

Esto nos da el sistema:
- $a + b + c = 0$
- $a + b = 0$
- $a = 1$

Resolviendo:
- De la tercera ecuación: $a = 1$
- Sustituyendo en la segunda: $1 + b = 0$, por lo que $b = -1$
- Sustituyendo en la primera: $1 + (-1) + c = 0$, por lo que $c = 0$

Por lo tanto, $u_3^* = (1, -1, 0)$.

### Resultado final

La base dual es:
$$\mathcal{B}^* = \{u_1^* = (0, 0, 1), u_2^* = (0, 1, -1), u_3^* = (1, -1, 0)\}$$

## Verificación

Comprobemos que se cumple la condición $u_i^*(u_j) = \delta_{ij}$:

- $u_1^*(u_1) = \langle (0,0,1), (1,1,1) \rangle = 0 \cdot 1 + 0 \cdot 1 + 1 \cdot 1 = 1$
- $u_1^*(u_2) = \langle (0,0,1), (1,1,0) \rangle = 0 \cdot 1 + 0 \cdot 1 + 1 \cdot 0 = 0$
- $u_1^*(u_3) = \langle (0,0,1), (1,0,0) \rangle = 0 \cdot 1 + 0 \cdot 0 + 1 \cdot 0 = 0$

- $u_2^*(u_1) = \langle (0,1,-1), (1,1,1) \rangle = 0 \cdot 1 + 1 \cdot 1 + (-1) \cdot 1 = 0$
- $u_2^*(u_2) = \langle (0,1,-1), (1,1,0) \rangle = 0 \cdot 1 + 1 \cdot 1 + (-1) \cdot 0 = 1$
- $u_2^*(u_3) = \langle (0,1,-1), (1,0,0) \rangle = 0 \cdot 1 + 1 \cdot 0 + (-1) \cdot 0 = 0$

- $u_3^*(u_1) = \langle (1,-1,0), (1,1,1) \rangle = 1 \cdot 1 + (-1) \cdot 1 + 0 \cdot 1 = 0$
- $u_3^*(u_2) = \langle (1,-1,0), (1,1,0) \rangle = 1 \cdot 1 + (-1) \cdot 1 + 0 \cdot 0 = 0$
- $u_3^*(u_3) = \langle (1,-1,0), (1,0,0) \rangle = 1 \cdot 1 + (-1) \cdot 0 + 0 \cdot 0 = 1$

Todas las condiciones se cumplen, lo que confirma que hemos encontrado correctamente la base dual.

## Conclusión y análisis geométrico

La base dual $\mathcal{B}^* = \{u_1^* = (0, 0, 1), u_2^* = (0, 1, -1), u_3^* = (1, -1, 0)\}$ tiene una interpretación geométrica interesante. Cada vector $u_i^*$ es perpendicular al subespacio generado por los vectores de la base original exceptuando $u_i$.

Por ejemplo:
- $u_1^*$ es perpendicular tanto a $u_2$ como a $u_3$, ya que $\langle u_1^*, u_2 \rangle = \langle u_1^*, u_3 \rangle = 0$
- $u_2^*$ es perpendicular tanto a $u_1$ como a $u_3$, ya que $\langle u_2^*, u_1 \rangle = \langle u_2^*, u_3 \rangle = 0$
- $u_3^*$ es perpendicular tanto a $u_1$ como a $u_2$, ya que $\langle u_3^*, u_1 \rangle = \langle u_3^*, u_2 \rangle = 0$

Esta propiedad hace que la base dual sea especialmente útil para proyectar vectores sobre los espacios generados por subconjuntos de la base original.

El código Python proporcionado visualiza:
1. La base original y la base dual en gráficos separados
2. Una representación que muestra la ortogonalidad entre los vectores duales y los subespacios correspondientes
3. Una verificación numérica de la propiedad $u_i^*(u_j) = \delta_{ij}$

Para ejecutar este código en Google Colab, simplemente copia y pega el código en una celda y ejecútalo. Esto generará las visualizaciones y la verificación de la base dual.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import matplotlib.colors as mcolors

# Definimos los vectores de la base original
u1 = np.array([1, 1, 1])
u2 = np.array([1, 1, 0])
u3 = np.array([1, 0, 0])

# Definimos los vectores de la base dual
u1_dual = np.array([0, 0, 1])
u2_dual = np.array([0, 1, -1])
u3_dual = np.array([1, -1, 0])

def create_plot():
    fig = plt.figure(figsize=(15, 12))

    # Primer subplot: Base original
    ax1 = fig.add_subplot(121, projection='3d')
    plot_base(ax1, u1, u2, u3, "Base Original", ['r', 'g', 'b'])

    # Segundo subplot: Base dual
    ax2 = fig.add_subplot(122, projection='3d')
    plot_base(ax2, u1_dual, u2_dual, u3_dual, "Base Dual", ['r', 'g', 'b'])

    plt.tight_layout()
    return fig

def plot_base(ax, v1, v2, v3, title, colors):
    # Origen
    origen = np.zeros(3)

    # Dibujamos los ejes
    max_range = np.max([np.max(np.abs(v1)), np.max(np.abs(v2)), np.max(np.abs(v3))]) * 1.5
    for i, s in enumerate(('x', 'y', 'z')):
        ax.plot([0, 0, 0], [0, 0, 0], [0, 0, 0], 'w')
        ax.plot([-max_range, max_range], [0, 0], [0, 0], 'k--', alpha=0.5)
        ax.plot([0, 0], [-max_range, max_range], [0, 0], 'k--', alpha=0.5)
        ax.plot([0, 0], [0, 0], [-max_range, max_range], 'k--', alpha=0.5)

    # Dibujamos los vectores
    ax.quiver(origen[0], origen[1], origen[2], v1[0], v1[1], v1[2], color=colors[0], arrow_length_ratio=0.1, linewidth=2, label=f'u1')
    ax.quiver(origen[0], origen[1], origen[2], v2[0], v2[1], v2[2], color=colors[1], arrow_length_ratio=0.1, linewidth=2, label=f'u2')
    ax.quiver(origen[0], origen[1], origen[2], v3[0], v3[1], v3[2], color=colors[2], arrow_length_ratio=0.1, linewidth=2, label=f'u3')

    # Dibujamos el paralelogramo generado por los vectores
    vertices = np.array([
        [0, 0, 0],
        v1,
        v1 + v2,
        v2,
        v3,
        v3 + v1,
        v3 + v1 + v2,
        v3 + v2
    ])

    # Definimos las caras del paralelogramo
    faces = [
        [vertices[0], vertices[1], vertices[2], vertices[3]],  # Base inferior
        [vertices[4], vertices[5], vertices[6], vertices[7]],  # Base superior
        [vertices[0], vertices[1], vertices[5], vertices[4]],  # Cara lateral 1
        [vertices[1], vertices[2], vertices[6], vertices[5]],  # Cara lateral 2
        [vertices[2], vertices[3], vertices[7], vertices[6]],  # Cara lateral 3
        [vertices[3], vertices[0], vertices[4], vertices[7]]   # Cara lateral 4
    ]

    # Añadimos el paralelogramo con transparencia
    collection = Poly3DCollection(faces, alpha=0.2, linewidths=1, edgecolors='gray')
    face_colors = [mcolors.to_rgba('cyan', alpha=0.2) for i in range(len(faces))]
    collection.set_facecolors(face_colors)
    ax.add_collection3d(collection)

    # Configuraciones adicionales
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title(title)
    ax.legend()

    # Ajustamos los límites
    ax.set_xlim([-max_range, max_range])
    ax.set_ylim([-max_range, max_range])
    ax.set_zlim([-max_range, max_range])

# Verificamos que la base dual cumple u_i^*(u_j) = delta_ij
def verify_dual_basis():
    print("Verificación de la base dual:")
    for i, ui_dual in enumerate([u1_dual, u2_dual, u3_dual], 1):
        for j, uj in enumerate([u1, u2, u3], 1):
            product = np.dot(ui_dual, uj)
            expected = 1 if i == j else 0
            print(f"u{i}*(u{j}) = {product}, Esperado: {expected}, ¿Correcto? {product == expected}")

# Crea una figura adicional que muestra la propiedad geométrica de la base dual
def show_orthogonality():
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Origen
    origen = np.zeros(3)

    # Dibujamos todos los vectores
    ax.quiver(origen[0], origen[1], origen[2], u1[0], u1[1], u1[2], color='r', arrow_length_ratio=0.1, linewidth=2, label='u1')
    ax.quiver(origen[0], origen[1], origen[2], u2[0], u2[1], u2[2], color='g', arrow_length_ratio=0.1, linewidth=2, label='u2')
    ax.quiver(origen[0], origen[1], origen[2], u3[0], u3[1], u3[2], color='b', arrow_length_ratio=0.1, linewidth=2, label='u3')

    ax.quiver(origen[0], origen[1], origen[2], u1_dual[0], u1_dual[1], u1_dual[2], color='orange', arrow_length_ratio=0.1, linewidth=2, label='u1*')
    ax.quiver(origen[0], origen[1], origen[2], u2_dual[0], u2_dual[1], u2_dual[2], color='purple', arrow_length_ratio=0.1, linewidth=2, label='u2*')
    ax.quiver(origen[0], origen[1], origen[2], u3_dual[0], u3_dual[1], u3_dual[2], color='brown', arrow_length_ratio=0.1, linewidth=2, label='u3*')

    # Configuraciones adicionales
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title('Visualización de la ortogonalidad entre vectores duales')
    ax.legend()

    max_range = 1.5
    ax.set_xlim([-max_range, max_range])
    ax.set_ylim([-max_range, max_range])
    ax.set_zlim([-max_range, max_range])

    return fig

# Ejecutamos las funciones
fig1 = create_plot()
fig2 = show_orthogonality()
verify_dual_basis()

# Para mostrar los gráficos en Colab, usa:
# plt.show()
```
