# Primer Teorema de Isomorfía para Espacios Vectoriales

## Definición Formal

Sea $T: V \rightarrow W$ una transformación lineal entre espacios vectoriales $V$ y $W$ sobre un campo $\mathbb{K}$. El Primer Teorema de Isomorfía establece que:

$$V/\ker(T) \cong \text{Im}(T)$$

donde:
- $V/\ker(T)$ es el espacio cociente de $V$ por el núcleo de $T$
- $\text{Im}(T)$ es la imagen de $T$
- $\cong$ denota un isomorfismo de espacios vectoriales

## Explicación Detallada

### Componentes Clave

1. **Transformación Lineal**: $T: V \rightarrow W$ es una aplicación que preserva las operaciones vectoriales:
   - $T(u + v) = T(u) + T(v)$ para todo $u, v \in V$
   - $T(\alpha v) = \alpha T(v)$ para todo $v \in V$ y $\alpha \in \mathbb{K}$

2. **Núcleo (Kernel)**: $\ker(T) = \{v \in V \mid T(v) = 0_W\}$
   - Es el conjunto de todos los vectores en $V$ que se mapean al vector cero en $W$
   - Es un subespacio de $V$

3. **Imagen**: $\text{Im}(T) = \{T(v) \mid v \in V\} \subseteq W$
   - Es el conjunto de todos los vectores en $W$ que son imagen de algún vector en $V$
   - Es un subespacio de $W$

4. **Espacio Cociente**: $V/\ker(T)$ consta de las clases de equivalencia $[v] = v + \ker(T)$
   - Cada clase $[v]$ contiene todos los vectores de la forma $v + n$ donde $n \in \ker(T)$
   - Dos vectores $u, v \in V$ están en la misma clase si y solo si $u - v \in \ker(T)$
   - Equivalentemente, $u$ y $v$ están en la misma clase si y solo si $T(u) = T(v)$

### El Isomorfismo Canónico

El teorema define un isomorfismo canónico $\varphi: V/\ker(T) \rightarrow \text{Im}(T)$ dado por:

$$\varphi([v]) = T(v)$$

Este mapeo está:
1. **Bien definido**: Si $[u] = [v]$, entonces $T(u) = T(v)$
2. **Es inyectivo**: Si $\varphi([u]) = \varphi([v])$, entonces $T(u) = T(v)$, lo que implica que $u - v \in \ker(T)$, por lo tanto $[u] = [v]$
3. **Es sobreyectivo**: Para cada $w \in \text{Im}(T)$, existe $v \in V$ tal que $T(v) = w$, por lo que $\varphi([v]) = w$
4. **Es lineal**: $\varphi(\alpha[u] + \beta[v]) = \varphi([\alpha u + \beta v]) = T(\alpha u + \beta v) = \alpha T(u) + \beta T(v) = \alpha\varphi([u]) + \beta\varphi([v])$

## Significado Geométrico

El teorema establece una correspondencia uno a uno entre:
- Las clases de equivalencia en $V/\ker(T)$ (cada una siendo un desplazamiento del núcleo $\ker(T)$)
- Los puntos en la imagen $\text{Im}(T)$

Podemos entender esto como una "compresión" del espacio $V$ a lo largo del núcleo $\ker(T)$, resultando en un espacio isomorfo a $\text{Im}(T)$.

## Descomposición Canónica

El Primer Teorema de Isomorfía nos permite descomponer el espacio $V$ como:

$$V \cong \ker(T) \oplus (V/\ker(T)) \cong \ker(T) \oplus \text{Im}(T)$$

donde:
- $\ker(T)$ representa la parte de $V$ que se "colapsa" a cero bajo $T$
- $V/\ker(T) \cong \text{Im}(T)$ representa la parte de $V$ que se "preserva" (aunque posiblemente distorsionada) bajo $T$

Esta descomposición es fundamental en muchas áreas del álgebra lineal y sus aplicaciones.

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Definir la transformación lineal T: R³ → R²
# Ejemplo: proyección sobre el plano xy
# T(x, y, z) = (x, y, 0)
def T(v):
    return np.array([v[0], v[1]])

# Crear una figura con dos subplots
fig = make_subplots(
    rows=1, cols=2,
    specs=[[{'type': 'scatter3d'}, {'type': 'scatter'}]],
    subplot_titles=('Espacio original R³ con kernel', 'Imagen T(R³) ≅ R²'),
    horizontal_spacing=0.05
)

# ---- Espacio original R³ ----
# Generar puntos en R³
x = y = z = np.linspace(-5, 5, 3)
X, Y, Z = np.meshgrid(x, y, z)
points = np.vstack([X.flatten(), Y.flatten(), Z.flatten()]).T

# Kernel de T: es el eje z
z_axis = np.array([[0, 0, -6], [0, 0, 6]])

# Complemento del kernel: representantes de las clases de equivalencia
# Eligimos el plano z=0 como representantes de V/ker(T)
xy_plane_x, xy_plane_y = np.meshgrid(np.linspace(-5, 5, 10), np.linspace(-5, 5, 10))
xy_plane_z = np.zeros_like(xy_plane_x)
xy_plane = np.vstack([xy_plane_x.flatten(), xy_plane_y.flatten(), xy_plane_z.flatten()]).T

# Generar algunos vectores ejemplo para clases de equivalencia
classes = []
colors = ['red', 'green', 'blue', 'purple', 'orange']
for i, (x, y) in enumerate([(2, 3), (-3, 1), (4, -2), (-1, -4), (0, 0)]):
    if i >= len(colors):
        break
    # Crear una línea paralela al kernel que pasa por el punto (x, y, 0)
    line = np.array([[x, y, -6], [x, y, 6]])
    classes.append((line, colors[i]))

# Añadir el plano z=0 como mesh para representar el complemento del kernel
fig.add_trace(
    go.Mesh3d(
        x=xy_plane[:, 0], y=xy_plane[:, 1], z=xy_plane[:, 2],
        opacity=0.5, color='lightblue',
        name='Complemento de ker(T)'
    ),
    row=1, col=1
)

# Añadir kernel (eje z)
fig.add_trace(
    go.Scatter3d(
        x=z_axis[:, 0], y=z_axis[:, 1], z=z_axis[:, 2],
        mode='lines',
        line=dict(color='red', width=6),
        name='ker(T) = eje z'
    ),
    row=1, col=1
)

# Añadir clases de equivalencia (líneas paralelas al kernel)
for i, (line, color) in enumerate(classes):
    fig.add_trace(
        go.Scatter3d(
            x=line[:, 0], y=line[:, 1], z=line[:, 2],
            mode='lines',
            line=dict(color=color, width=4),
            name=f'Clase [{line[0, 0]}, {line[0, 1]}, z]'
        ),
        row=1, col=1
    )

    # Añadir punto en el plano xy donde intersecta con la clase
    fig.add_trace(
        go.Scatter3d(
            x=[line[0, 0]], y=[line[0, 1]], z=[0],
            mode='markers',
            marker=dict(color=color, size=8),
            name=f'Representante ({line[0, 0]}, {line[0, 1]}, 0)'
        ),
        row=1, col=1
    )

# ---- Imagen en R² ----
# La imagen es el plano xy, que visualizamos como R²
# Crear límites del plano R²
x_range = np.array([-5, 5])
y_range = np.array([-5, 5])
img_boundary = np.array([
    [x_range[0], y_range[0]],
    [x_range[1], y_range[0]],
    [x_range[1], y_range[1]],
    [x_range[0], y_range[1]],
    [x_range[0], y_range[0]]
])

# Agregar límites del plano R²
fig.add_trace(
    go.Scatter(
        x=img_boundary[:, 0], y=img_boundary[:, 1],
        mode='lines',
        line=dict(color='black', width=2),
        name='Im(T) ≅ R²'
    ),
    row=1, col=2
)

# Agregar puntos para las imágenes de representantes
for i, (line, color) in enumerate(classes):
    # Aplicar T a un punto de la clase
    img_point = T(line[0])
    fig.add_trace(
        go.Scatter(
            x=[img_point[0]], y=[img_point[1]],
            mode='markers',
            marker=dict(color=color, size=10),
            name=f'T({line[0, 0]}, {line[0, 1]}, 0) = ({img_point[0]}, {img_point[1]})'
        ),
        row=1, col=2
    )

# Configurar aspecto
fig.update_layout(
    title='Primer Teorema de Isomorfía: V/ker(T) ≅ Im(T)',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='cube'
    ),
    xaxis_title='X',
    yaxis_title='Y',
    width=1000,
    height=600,
    legend=dict(x=1.1, y=0.5)
)

fig.update_scenes(
    camera=dict(
        eye=dict(x=1.5, y=-1.5, z=1)
    )
)

fig.show()
````

# Ejemplo del Primer Teorema de Isomorfía en R³

Vamos a ilustrar el Primer Teorema de Isomorfía con un ejemplo concreto y sencillo en R³.

## Transformación Lineal

Consideremos la transformación lineal $T: \mathbb{R}^3 \rightarrow \mathbb{R}^2$ definida por:

$$T(x, y, z) = (x, y)$$

Esta transformación es simplemente una proyección que "olvida" la coordenada $z$.

## Análisis de la Transformación

### 1. Núcleo (Kernel)

El núcleo de $T$ es:

$$\ker(T) = \{(x, y, z) \in \mathbb{R}^3 \mid T(x, y, z) = (0, 0)\} = \{(0, 0, z) \mid z \in \mathbb{R}\}$$

Geométricamente, $\ker(T)$ es el eje $z$. Es un subespacio de dimensión 1 de $\mathbb{R}^3$.

### 2. Imagen

La imagen de $T$ es:

$$\text{Im}(T) = \{(x, y) \mid (x, y) = T(x, y, z) \text{ para algún } (x, y, z) \in \mathbb{R}^3\} = \mathbb{R}^2$$

Es decir, la imagen es todo $\mathbb{R}^2$, un espacio de dimensión 2.

### 3. Espacio Cociente

El espacio cociente $\mathbb{R}^3/\ker(T)$ consiste en clases de equivalencia de la forma:

$$[(x, y, z)] = \{(x, y, z') \mid z' \in \mathbb{R}\}$$

Cada clase es una línea paralela al eje $z$ que pasa por el punto $(x, y, z)$. Todas las líneas con las mismas coordenadas $x$ y $y$ representan la misma clase.

## Verificación del Teorema

Según el Primer Teorema de Isomorfía, debemos tener:

$$\mathbb{R}^3/\ker(T) \cong \text{Im}(T) = \mathbb{R}^2$$

El isomorfismo canónico $\varphi: \mathbb{R}^3/\ker(T) \rightarrow \mathbb{R}^2$ está dado por:

$$\varphi([(x, y, z)]) = T(x, y, z) = (x, y)$$

### Comprobación:

1. **Bien definido**: Si $(x_1, y_1, z_1)$ y $(x_2, y_2, z_2)$ están en la misma clase, entonces $(x_1, y_1) = (x_2, y_2)$, por lo que $T(x_1, y_1, z_1) = T(x_2, y_2, z_2)$.

2. **Inyectivo**: Si $\varphi([(x_1, y_1, z_1)]) = \varphi([(x_2, y_2, z_2)])$, entonces $(x_1, y_1) = (x_2, y_2)$, lo que implica que $(x_1, y_1, z_1)$ y $(x_2, y_2, z_2)$ están en la misma clase.

3. **Sobreyectivo**: Para cualquier $(x, y) \in \mathbb{R}^2$, el vector $(x, y, 0) \in \mathbb{R}^3$ satisface $T(x, y, 0) = (x, y)$.

4. **Lineal**: Es claro que $\varphi$ preserva las operaciones vectoriales.

## Descomposición Canónica

La descomposición canónica nos dice que:

$$\mathbb{R}^3 \cong \ker(T) \oplus (\mathbb{R}^3/\ker(T)) \cong \ker(T) \oplus \mathbb{R}^2$$

En nuestro ejemplo, podemos visualizar esto como la descomposición de $\mathbb{R}^3$ en:
- El eje $z$ (el kernel de dimensión 1)
- El plano $xy$ (un complemento del kernel, isomorfo a $\mathbb{R}^2$)

Cualquier vector $(x, y, z) \in \mathbb{R}^3$ puede descomponerse de manera única como:

$$(x, y, z) = (0, 0, z) + (x, y, 0)$$

donde $(0, 0, z) \in \ker(T)$ y $(x, y, 0)$ es un representante de la clase $[(x, y, z)]$ en el espacio cociente.

## Interpretación Geométrica

En la visualización, podemos observar:

1. En el espacio original $\mathbb{R}^3$:
   - El eje $z$ en rojo representa el núcleo de $T$
   - El plano $z=0$ (en azul claro) representa un conjunto de representantes del espacio cociente
   - Las líneas verticales de colores representan clases de equivalencia individuales

2. En el espacio imagen $\mathbb{R}^2$:
   - Cada punto coloreado corresponde a la imagen de una clase completa
   - Hay una correspondencia uno a uno entre puntos en $\mathbb{R}^2$ y clases en $\mathbb{R}^3/\ker(T)$

Esta visualización muestra claramente cómo $T$ "colapsa" cada línea paralela al eje $z$ (una clase de equivalencia) en un único punto del plano $\mathbb{R}^2$.

# EXPLICACION ALTERNATIVA Y CÓDIGO ALTERNATIVO

El Primer Teorema del Isomorfismo (también conocido como Teorema de la Dimensión) establece que si T: V → W es una transformación lineal, entonces:

dim(V) = dim(Ker(T)) + dim(Im(T))

Donde:

V y W son espacios vectoriales.
Ker(T) es el núcleo de T (el conjunto de vectores en V que se mapean al vector cero en W).
Im(T) es la imagen de T (el conjunto de todos los vectores en W que pueden ser obtenidos como la salida de T para algún vector en V).
Ejemplo:

Consideremos V = R³ y W = R². Definimos una transformación lineal T: R³ → R² como:

T(x, y, z) = (x + y, y - z)

1. Calcular el Núcleo de T (Ker(T)):

Para encontrar Ker(T), necesitamos resolver la ecuación T(x, y, z) = (0, 0), lo que implica:

x + y = 0
y - z = 0
De la primera ecuación, tenemos y = -x.  Sustituyendo en la segunda ecuación, obtenemos -x - z = 0, o z = -x.  Por lo tanto, cualquier vector de la forma (x, -x, -x) está en Ker(T).

Una base para Ker(T) es {(1, -1, -1)}.  Así, dim(Ker(T)) = 1.

2. Calcular la Imagen de T (Im(T)):

La imagen de T es el conjunto de todos los vectores (a, b) en R² que pueden ser escritos como (x + y, y - z) para algunos valores de x, y, z.  Observamos que podemos obtener cualquier vector (a, b) en R² eligiendo:

x = a - b
y = 0
z = -b
Por lo tanto, Im(T) es todo el espacio R². Una base para Im(T) es {(1, 0), (0, 1)}.  Así, dim(Im(T)) = 2.

3. Verificar el Teorema:

Tenemos:

dim(V) = dim(R³) = 3
dim(Ker(T)) = 1
dim(Im(T)) = 2
El teorema establece que dim(V) = dim(Ker(T)) + dim(Im(T)).  En nuestro caso, 3 = 1 + 2, lo cual se cumple.  Por lo tanto, hemos demostrado el Primer Teorema del Isomorfismo en este ejemplo.

````python
import numpy as np
import plotly.graph_objects as go

# Definición de la transformación lineal T
def T(x, y, z):
  return x + y, y - z

# Generación de puntos en R³ para visualizar
num_points = 50
np.random.seed(42) # Para reproducibilidad
x = np.random.uniform(-3, 3, num_points)
y = np.random.uniform(-3, 3, num_points)
z = np.random.uniform(-3, 3, num_points)

# Aplicación de la transformación T a los puntos
a, b = T(x, y, z)

# Visualización en Plotly
fig = go.Figure()

# Scatter plot de los puntos originales en R³ (proyección en el plano XY para simplificar)
fig.add_trace(go.Scatter3d(
    x=x,
    y=y,
    z=np.zeros_like(x), # Proyectamos sobre el plano xy
    mode='markers',
    marker=dict(size=5, color='blue'),
    name='Puntos en R³'
))

# Scatter plot de los puntos transformados en R²
fig.add_trace(go.Scatter(
    x=a,
    y=b,
    mode='markers',
    marker=dict(size=5, color='red'),
    name='Puntos en Im(T)'
))

# Líneas que representan el núcleo de T (aproximación visual)
# El núcleo es la línea definida por (x, -x, -x).  Visualizamos una sección.
t_values = np.linspace(-3, 3, 20)
kernel_x = t_values
kernel_y = -t_values
kernel_z = -t_values

fig.add_trace(go.Scatter3d(
    x=kernel_x,
    y=kernel_y,
    z=kernel_z,
    mode='lines',
    line=dict(color='green'),
    name='Aproximación del Núcleo (Ker(T))'
))

fig.update_layout(
    title='Ilustración del Primer Teorema del Isomorfismo',
    scene = dict(
        xaxis_title='x',
        yaxis_title='y',
        zaxis_title='z'
    )
)

fig.show()
````

# OTRO EJEMPLO



### 1. Ejemplo detallado con matrices
He escogido la transformación lineal T: R³ → R² definida por:
```
T(x, y, z) = (x + z, y - z)
```

### 2. Análisis paso a paso
- Calculé la matriz de T en la base canónica:
  ```
  [T]_E^E = [ 1  0  1 ]
            [ 0  1 -1 ]
  ```
- Encontré el núcleo ker(T) = span{(-1, 1, 1)}
- Construí una base adaptada B = {(-1, 1, 1), (1, 0, 0), (0, 1, 0)}
- Calculé la matriz de cambio de base y la matriz de T en esta base adaptada

### 3. Demostración del isomorfismo
La matriz de T en la base adaptada tiene la forma:
```
[T]_B^E = [ 0  1  0 ]
          [ 0  0  1 ]
```

Esta forma matricial muestra exactamente la descomposición:
- Primera columna nula: corresponde al vector del núcleo
- Las otras dos columnas forman una matriz identidad 2×2: representan el isomorfismo entre V/ker(T) y Im(T)

### 4. Visualización geométrica
La visualización ilustra tres aspectos clave:
- Espacio original R³ con vectores base y clases de equivalencia (líneas paralelas al kernel)
- Descomposición R³ = ker(T) ⊕ W donde W es el complemento del núcleo
- Imagen en R² mostrando cómo las clases se mapean isomórficamente

Cada clase de equivalencia (línea paralela al kernel) se mapea a un único punto en R², estableciendo el isomorfismo V/ker(T) ≅ Im(T).

### 5. Interpretación matricial
El cambio de base adaptada revela que:
- La transformación T puede expresarse como la proyección al complemento del núcleo seguida de un isomorfismo
- La matriz en forma normal [0 | I₂] muestra exactamente la estructura del Primer Teorema de Isomorfía
- Los vectores v₂ = (1,0,0) y v₃ = (0,1,0) forman una base del complemento del núcleo, y sus imágenes T(v₂) = (1,0) y T(v₃) = (0,1) forman una base de la imagen

Este enfoque matricial es especialmente útil para comprender cómo el cambio de base adecuado revela la estructura del isomorfismo entre el espacio cociente y la imagen de la transformación lineal.

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Definir la transformación lineal T: R³ → R²
# T(x, y, z) = (x + z, y - z)
def T(v):
    return np.array([v[0] + v[2], v[1] - v[2]])

# Crear una figura con tres subplots
fig = make_subplots(
    rows=1, cols=3,
    specs=[[{'type': 'scatter3d'}, {'type': 'scatter3d'}, {'type': 'scatter'}]],
    subplot_titles=('Espacio original R³', 'Descomposición R³ = ker(T) ⊕ W', 'Imagen T(R³) ≅ R²'),
    horizontal_spacing=0.05
)

# Definir nuestras bases
# Base del kernel: v1 = (-1, 1, 1)
# Base complementaria: v2 = (1, 0, 0), v3 = (0, 1, 0)
v1 = np.array([-1, 1, 1])
v2 = np.array([1, 0, 0])
v3 = np.array([0, 1, 0])

# Crear el origen
origin = np.zeros(3)

# --- Primer gráfico: Espacio original R³ con vectores base ---
# Vectores de la base B
vectors = [v1, v2, v3]
colors = ['red', 'green', 'blue']
names = ['v₁ = (-1, 1, 1) [kernel]', 'v₂ = (1, 0, 0)', 'v₃ = (0, 1, 0)']

for i, (v, color, name) in enumerate(zip(vectors, colors, names)):
    fig.add_trace(
        go.Scatter3d(
            x=[origin[0], v[0]], y=[origin[1], v[1]], z=[origin[2], v[2]],
            mode='lines+markers',
            line=dict(color=color, width=5),
            marker=dict(size=[0, 8]),
            name=name
        ),
        row=1, col=1
    )

# Crear algunas clases de equivalencia (líneas paralelas al kernel)
class_points = [(2, 1, 0), (0, 2, 0), (-1, -1, 0), (1, -2, 0)]
class_colors = ['purple', 'orange', 'cyan', 'magenta']

for i, (point, color) in enumerate(zip(class_points, class_colors)):
    # Crear punto de referencia
    point = np.array(point)

    # Crear línea en la dirección del kernel que pasa por el punto
    t = np.linspace(-2, 2, 50)
    line_points = np.array([point + t_i * v1 for t_i in t])

    fig.add_trace(
        go.Scatter3d(
            x=line_points[:, 0], y=line_points[:, 1], z=line_points[:, 2],
            mode='lines',
            line=dict(color=color, width=4),
            name=f'Clase de equivalencia [{point[0]}, {point[1]}, {point[2]}] + t·v₁'
        ),
        row=1, col=1
    )

    # Agregar el punto
    fig.add_trace(
        go.Scatter3d(
            x=[point[0]], y=[point[1]], z=[point[2]],
            mode='markers',
            marker=dict(color=color, size=6),
            name=f'Representante ({point[0]}, {point[1]}, {point[2]})'
        ),
        row=1, col=1
    )

# --- Segundo gráfico: Descomposición R³ = ker(T) ⊕ W ---
# Dibujar el kernel
t = np.linspace(-3, 3, 100)
kernel_line = np.array([t_i * v1 for t_i in t])

fig.add_trace(
    go.Scatter3d(
        x=kernel_line[:, 0], y=kernel_line[:, 1], z=kernel_line[:, 2],
        mode='lines',
        line=dict(color='red', width=6),
        name='ker(T) = span{(-1, 1, 1)}'
    ),
    row=1, col=2
)

# Dibujar el complemento W = span{v2, v3}
# Crear malla para el plano
u = np.linspace(-3, 3, 10)
v = np.linspace(-3, 3, 10)
u_grid, v_grid = np.meshgrid(u, v)
x = u_grid
y = v_grid
z = np.zeros_like(u_grid)

fig.add_trace(
    go.Surface(
        x=x, y=y, z=z,
        colorscale=[[0, 'lightblue'], [1, 'lightblue']],
        opacity=0.6,
        showscale=False,
        name='W = span{(1,0,0), (0,1,0)}'
    ),
    row=1, col=2
)

# Agregar vectores base para W
for i, (v, color, name) in enumerate(zip([v2, v3], ['green', 'blue'], ['v₂ = (1, 0, 0)', 'v₃ = (0, 1, 0)'])):
    fig.add_trace(
        go.Scatter3d(
            x=[origin[0], v[0]], y=[origin[1], v[1]], z=[origin[2], v[2]],
            mode='lines+markers',
            line=dict(color=color, width=5),
            marker=dict(size=[0, 8]),
            name=name
        ),
        row=1, col=2
    )

# Agregar las mismas clases de equivalencia del primer gráfico
for i, (point, color) in enumerate(zip(class_points, class_colors)):
    point = np.array(point)

    # Línea en la dirección del kernel que pasa por el punto
    t = np.linspace(-2, 2, 50)
    line_points = np.array([point + t_i * v1 for t_i in t])

    fig.add_trace(
        go.Scatter3d(
            x=line_points[:, 0], y=line_points[:, 1], z=line_points[:, 2],
            mode='lines',
            line=dict(color=color, width=4),
            showlegend=False
        ),
        row=1, col=2
    )

    # Agregar el punto
    fig.add_trace(
        go.Scatter3d(
            x=[point[0]], y=[point[1]], z=[point[2]],
            mode='markers',
            marker=dict(color=color, size=6),
            showlegend=False
        ),
        row=1, col=2
    )

# --- Tercer gráfico: Imagen T(R³) ≅ R² ---
# Límites del plano R²
x_range = np.array([-3, 3])
y_range = np.array([-3, 3])
img_boundary = np.array([
    [x_range[0], y_range[0]],
    [x_range[1], y_range[0]],
    [x_range[1], y_range[1]],
    [x_range[0], y_range[1]],
    [x_range[0], y_range[0]]
])

# Agregar límites del plano R²
fig.add_trace(
    go.Scatter(
        x=img_boundary[:, 0], y=img_boundary[:, 1],
        mode='lines',
        line=dict(color='black', width=2),
        name='Im(T) ≅ R²'
    ),
    row=1, col=3
)

# Dibujar ejes del plano
fig.add_trace(
    go.Scatter(
        x=[-3, 3], y=[0, 0],
        mode='lines',
        line=dict(color='gray', width=1),
        showlegend=False
    ),
    row=1, col=3
)

fig.add_trace(
    go.Scatter(
        x=[0, 0], y=[-3, 3],
        mode='lines',
        line=dict(color='gray', width=1),
        showlegend=False
    ),
    row=1, col=3
)

# Agregar imágenes de los vectores base
T_v2 = T(v2)  # (1, 0)
T_v3 = T(v3)  # (0, 1)

fig.add_trace(
    go.Scatter(
        x=[0, T_v2[0]], y=[0, T_v2[1]],
        mode='lines+markers',
        line=dict(color='green', width=3),
        marker=dict(size=[0, 6]),
        name='T(v₂) = (1, 0)'
    ),
    row=1, col=3
)

fig.add_trace(
    go.Scatter(
        x=[0, T_v3[0]], y=[0, T_v3[1]],
        mode='lines+markers',
        line=dict(color='blue', width=3),
        marker=dict(size=[0, 6]),
        name='T(v₃) = (0, 1)'
    ),
    row=1, col=3
)

# Agregar imágenes de los puntos representantes de clases
for i, (point, color) in enumerate(zip(class_points, class_colors)):
    # Aplicar T al punto
    img_point = T(point)
    fig.add_trace(
        go.Scatter(
            x=[img_point[0]], y=[img_point[1]],
            mode='markers',
            marker=dict(color=color, size=8),
            name=f'T({point}) = ({img_point[0]}, {img_point[1]})'
        ),
        row=1, col=3
    )

# Configurar aspecto de los gráficos
fig.update_layout(
    title='Primer Teorema de Isomorfía mediante Matrices: R³/ker(T) ≅ Im(T)',
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='cube',
        camera=dict(
            eye=dict(x=1.5, y=-1.5, z=1)
        )
    ),
    scene2=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z',
        aspectmode='cube',
        camera=dict(
            eye=dict(x=1.5, y=-1.5, z=1)
        )
    ),
    xaxis_title='X',
    yaxis_title='Y',
    width=1200,
    height=600,
    legend=dict(x=1.1, y=0.5)
)

# Mostrar la figura
fig.show()
````


# Demostración del Primer Teorema de Isomorfismo en R³

## Introducción

El primer teorema de isomorfismo establece que si $T: V \rightarrow W$ es una aplicación lineal y $V$ es un espacio vectorial finitamente generado, entonces:

$$V/\ker(T) \cong \text{Im}(T)$$

Donde $V/\ker(T)$ representa el espacio cociente de $V$ por el núcleo de $T$, e $\text{Im}(T)$ representa la imagen de $T$.

En esta demostración, consideraremos el caso específico de $V = \mathbb{R}^3$ y demostraremos el teorema construyendo explícitamente las matrices de cambio de base en la descomposición canónica.

## Definición de la Aplicación Lineal

Sea $T: \mathbb{R}^3 \rightarrow \mathbb{R}^3$ una aplicación lineal isomorfa. Para hacer esta demostración concreta, consideremos la aplicación definida por la matriz:

$$A = \begin{pmatrix}
1 & 2 & 1 \\
0 & 1 & 1 \\
1 & 0 & 2
\end{pmatrix}$$

Esta es una matriz inversible (su determinante es 3), por lo que $T$ es un isomorfismo. En este caso particular, como $T$ es un isomorfismo, tenemos que $\ker(T) = \{0\}$ y $\text{Im}(T) = \mathbb{R}^3$.

## Caso General con $\ker(T) \neq \{0\}$

Para ilustrar mejor el teorema, consideremos una aplicación lineal que no sea un isomorfismo:

$$B = \begin{pmatrix}
1 & 0 & 2 \\
0 & 1 & 3 \\
1 & 1 & 5
\end{pmatrix}$$

Esta matriz tiene rango 2, por lo que $\dim(\ker(T)) = 1$ y $\dim(\text{Im}(T)) = 2$.

### Paso 1: Encontrar el Núcleo

Para encontrar $\ker(T)$, resolvemos la ecuación $B\vec{x} = \vec{0}$:

$$\begin{pmatrix}
1 & 0 & 2 \\
0 & 1 & 3 \\
1 & 1 & 5
\end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \\ 0 \end{pmatrix}$$

Reduciendo el sistema por eliminación gaussiana:

$$\begin{pmatrix}
1 & 0 & 2 \\
0 & 1 & 3 \\
0 & 0 & 0
\end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \\ 0 \end{pmatrix}$$

Obtenemos:
- $x_1 + 2x_3 = 0 \Rightarrow x_1 = -2x_3$
- $x_2 + 3x_3 = 0 \Rightarrow x_2 = -3x_3$

Por lo tanto, $\ker(T) = \{\lambda \cdot (-2, -3, 1) : \lambda \in \mathbb{R}\}$

### Paso 2: Construir una Base para $\mathbb{R}^3$ que incluya $\ker(T)$

Tomemos $\vec{v}_1 = (-2, -3, 1)$ como base para $\ker(T)$. Necesitamos completar esta base para $\mathbb{R}^3$. Podemos elegir:

$\vec{v}_2 = (1, 0, 0)$
$\vec{v}_3 = (0, 1, 0)$

Verificamos que estos tres vectores son linealmente independientes, por lo que forman una base para $\mathbb{R}^3$.

### Paso 3: Matriz de Cambio de Base de la Canónica a la Base $\{\vec{v}_1, \vec{v}_2, \vec{v}_3\}$

La matriz de cambio de base $P$ de la base canónica a nuestra nueva base está formada por los vectores de la nueva base como columnas:

$$P = \begin{pmatrix}
-2 & 1 & 0 \\
-3 & 0 & 1 \\
1 & 0 & 0
\end{pmatrix}$$

### Paso 4: Expresar $T$ en la Nueva Base

La matriz de $T$ en la nueva base es:

$$B' = P^{-1} \cdot B \cdot P$$

Primero calculamos $P^{-1}$:

$$P^{-1} = \frac{1}{\det(P)}\text{adj}(P)$$

Donde $\det(P) = -2 \cdot 1 \cdot 0 - (-3) \cdot 0 \cdot 0 - 1 \cdot 1 \cdot 1 - 0 \cdot 0 \cdot (-2) - 1 \cdot (-3) \cdot 0 - 0 \cdot 1 \cdot (-3) = -1$

Y la matriz adjunta es:

$$\text{adj}(P) = \begin{pmatrix}
0 & 0 & -1 \\
-3 & 1 & 3 \\
-2 & 0 & 2
\end{pmatrix}^T = \begin{pmatrix}
0 & -3 & -2 \\
0 & 1 & 0 \\
-1 & 3 & 2
\end{pmatrix}$$

Por lo tanto:

$$P^{-1} = \begin{pmatrix}
0 & 3 & 2 \\
0 & -1 & 0 \\
1 & -3 & -2
\end{pmatrix}$$

Ahora calculamos $B' = P^{-1} \cdot B \cdot P$:

$$B' = \begin{pmatrix}
0 & 3 & 2 \\
0 & -1 & 0 \\
1 & -3 & -2
\end{pmatrix} \cdot \begin{pmatrix}
1 & 0 & 2 \\
0 & 1 & 3 \\
1 & 1 & 5
\end{pmatrix} \cdot \begin{pmatrix}
-2 & 1 & 0 \\
-3 & 0 & 1 \\
1 & 0 & 0
\end{pmatrix}$$

Después de realizar los cálculos matriciales, obtenemos:

$$B' = \begin{pmatrix}
0 & 0 & 0 \\
0 & 1 & 3 \\
0 & 2 & 6
\end{pmatrix}$$

Observamos que la primera columna está compuesta solo por ceros, lo que refleja el hecho de que $\vec{v}_1 \in \ker(T)$.

### Paso 5: Descomposición Canónica

La matriz $B'$ tiene la forma:

$$B' = \begin{pmatrix}
0 & * & * \\
0 & * & * \\
0 & * & *
\end{pmatrix}$$

Esto muestra la descomposición canónica según el primer teorema de isomorfismo. La primera columna representa el espacio $\ker(T)$, mientras que las columnas 2 y 3 representan un complemento de $\ker(T)$ en $\mathbb{R}^3$.

Esta estructura de la matriz ilustra el isomorfismo $\mathbb{R}^3/\ker(T) \cong \text{Im}(T)$. Los vectores del cociente $\mathbb{R}^3/\ker(T)$ se pueden representar por los vectores en el complemento de $\ker(T)$, que a través de $T$ se mapean isomórficamente a $\text{Im}(T)$.

## Matriz de la Proyección Canónica

La proyección canónica $\pi: \mathbb{R}^3 \rightarrow \mathbb{R}^3/\ker(T)$ se puede representar por la matriz:

$$\Pi = \begin{pmatrix}
0 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{pmatrix}$$

en la base $\{\vec{v}_1, \vec{v}_2, \vec{v}_3\}$.

## Matriz del Isomorfismo Inducido

El isomorfismo inducido $\overline{T}: \mathbb{R}^3/\ker(T) \rightarrow \text{Im}(T)$ se puede representar por la matriz:

$$\overline{T} = \begin{pmatrix}
1 & 3 \\
2 & 6
\end{pmatrix}$$

Observamos que esta matriz tiene rango 1, lo que refleja una dependencia lineal en la representación elegida.

## Conclusión

Hemos demostrado explícitamente el primer teorema de isomorfismo para una aplicación lineal en $\mathbb{R}^3$, construyendo las matrices de cambio de base y mostrando la descomposición canónica. La matriz de la aplicación lineal en la base adaptada al núcleo muestra claramente la estructura del isomorfismo entre el espacio cociente $V/\ker(T)$ y la imagen $\text{Im}(T)$.

Esta demostración ilustra cómo, al elegir una base adecuada que incluya una base del núcleo, podemos visualizar geométricamente el primer teorema de isomorfismo a través de las matrices de cambio de base y la descomposición canónica.
